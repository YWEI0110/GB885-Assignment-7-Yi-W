##World Health

```python
#Import the Pandas and NumPy modules
import pandas as pd
import numpy as np
```

```python
#read csv file of JPG
url="https://raw.githubusercontent.com/katiegaertner/ML-Fundamentals/main/world-health-data-numbers.csv"
df_health=pd.read_csv(url)
```

```python
#Check the dimensions of the dataset
df_health.shape
```

```python
#Preview the dataframe
df_health.head()
```

###Update code

```python
#set Abbreviation as index
df_health = df_health.set_index("Abbreviation")
df_health.head()
```

```python
#Search nation index value "DZ"
df_health.loc["DZ"]
```

###Update code

```python
#drop Vatican City (abbreviation 'VA') from DataFrame.
df_health = df_health.drop("VA")
df_health.shape
```

```python
#Search the official language of the nation on the 186th row
df_health.sort_values(by="Abbreviation")
df_health.iloc[185][["Country", "Official Language"]]
```

```python
#Sort infant mortality from highest to lowest
df_health = df_health.sort_values("Infant mortality_per_1k", ascending=False)
df_health.head(10)
```

###Create a dataframe

```python
#Create a dataframe for Spanish-speaking nations
spanish_speaking_countries = df_health[df_health["Official Language"] == "Spanish"]

#Check the number of rows and columns of new dataframe
spanish_speaking_countries.shape
```

```python
#calculate mean of infant mortality in new dataframe
round(spanish_speaking_countries["Infant mortality_per_1k"].mean(), 2)
```

```python
#Calculate median maternal mortality in countries whose official language is Spanish
round(spanish_speaking_countries["Maternal_Mortality_per_100k"].median(), 2)
```

```python
#Create a new column for GDP per capita
df_health["GDP_per_capita"] = df_health["GDP"] / df_health["Population"]

# Sort from highest to lowest GDP per capita
gdp_per_capita_sorted = df_health.sort_values("GDP_per_capita", ascending=False)

#find nation with fourth-highest GDP per capita.
gdp_per_capita_sorted.iloc[3]["Country"]
```

```python
#Find the maternal mortality rate for Switzerland
round(gdp_per_capita_sorted.iloc[3]["Maternal_Mortality_per_100k"], 2)
```

```python
#Calculate the average maternal mortality rate per 100,000 for all nations
round(df_health["Maternal_Mortality_per_100k"].mean(), 2)
```

```python
# Find the country with the lowest infant mortality and its out-of-pocket health care percentage

lowest_infant_mortality = df_health.sort_values("Infant mortality_per_1k", ascending=True).iloc[0]

round(lowest_infant_mortality["oop_health_exp"] * 100, 2)

```

```python
# Get the out-of-pocket health care percentage for the country with the highest reported infant mortality

highest_infant_mortality = df_health.dropna(subset=["Infant mortality_per_1k"]) \
    .sort_values("Infant mortality_per_1k", ascending=False) \
    .iloc[0]

round(highest_infant_mortality["oop_health_exp"] * 100, 2)
```
